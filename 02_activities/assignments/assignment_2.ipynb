  ## Examples

  ### Example 1

  Input: `lst = [0, 2]`

  Output: [1]

  ### Example 2

  Input: `lst = [5, 0, 1]`

  Output: [2, 3, 4]

  ### Example 3

  Input: `lst = [6, 8, 2, 3, 5, 7, 0, 1, 10]`

  Output: [4, 9]

</details>

#### Starter Code for Question 3


```python
from typing import List

def missing_num(nums: List) -> int:
  # TODO
    n = max(nums)
    full_set = set(range(n + 1))
    missing = list(full_set - set(nums))
    return sorted(missing) if missing else -1

# Example usage:
nums = [0, 2]
print(missing_num(nums))  # Output: 1

nums = [5, 0, 1]
print(missing_num(nums))  # Output: 2

nums = [6, 8, 2, 3, 5, 7, 0, 1, 10]
print(missing_num(nums))  # Output: 3
```

```python
# Your answer here
from typing import List

def first_duplicate(nums: List[int]):
    seen = set()
    for n in nums:
        if n in seen:
            return n
        seen.add(n)
    return -1
```
